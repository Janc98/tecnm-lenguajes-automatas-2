````python
%%writefile api_sesion10.h
/* api_sesion10.h - contrato de la Sesion 10
 * Mini-evaluador de expresiones (Hito 1)
 *
 * Esta cabecera es el contrato: el harness y tu implementacion
 * deben incluirla. Si cambias una firma, el harness dejara de compilar.
 */
#ifndef API_SESION10_H
#define API_SESION10_H

long evaluar_expresion(const char *fuente, int *error);

#endif /* API_SESION10_H */
````

````python
%%writefile artesano_sesion10_c.c

/*
 * artesano_sesion10_c.c
 * Fase 1: El Artesano
 * Mini-evaluador de expresiones
 */

#include "api_sesion10.h"
#include <ctype.h>
#include <stddef.h>

/* Variables para leer la expresion */
static const char *posicion;
static int fallo;

/* Declaracion de funciones */
static long expresion(void);

/* Ignorar espacios */
static void espacios(void) {
    while (isspace((unsigned char)*posicion)) {
        posicion++;
    }
}

/* Leer numeros y parentesis */
static long numero(void) {
    long valor = 0;
    int signo = 1;

    espacios();

    if (*posicion == '-') {
        signo = -1;
        posicion++;
    } else if (*posicion == '+') {
        posicion++;
    }

    espacios();

    if (*posicion == '(') {
        posicion++;
        valor = expresion();
        espacios();

        if (*posicion != ')') {
            fallo = 2;
            return 0;
        }

        posicion++;
        return signo * valor;
    }

    if (!isdigit((unsigned char)*posicion)) {
        fallo = 2;
        return 0;
    }

    while (isdigit((unsigned char)*posicion)) {
        valor = valor * 10 + (*posicion - '0');
        posicion++;
    }

    return signo * valor;
}

/* Resolver multiplicacion y division */
static long termino(void) {
    long valor = numero();

    while (fallo == 0) {
        long segundo;

        espacios();

        if (*posicion == '*') {
            posicion++;
            segundo = numero();

            if (fallo == 0) {
                valor *= segundo;
            }
        } else if (*posicion == '/') {
            posicion++;
            segundo = numero();

            if (fallo == 0) {
                if (segundo == 0) {
                    fallo = 1;
                    return 0;
                }

                valor /= segundo;
            }
        } else {
            break;
        }
    }

    return valor;
}

/* Resolver suma y resta */
static long expresion(void) {
    long valor = termino();

    while (fallo == 0) {
        long segundo;

        espacios();

        if (*posicion == '+') {
            posicion++;
            segundo = termino();

            if (fallo == 0) {
                valor += segundo;
            }
        } else if (*posicion == '-') {
            posicion++;
            segundo = termino();

            if (fallo == 0) {
                valor -= segundo;
            }
        } else {
            break;
        }
    }

    return valor;
}

/* Funcion solicitada por el contrato */
long evaluar_expresion(const char *fuente, int *error) {
    long resultado = 0;

    fallo = 0;

    if (error != NULL) {
        *error = 0;
    }

    if (fuente == NULL) {
        fallo = 2;
    } else {
        posicion = fuente;

        resultado = expresion();

        espacios();

        if (*posicion != '\0' && fallo == 0) {
            fallo = 2;
        }
    }

    if (error != NULL) {
        *error = fallo;
    }

    if (fallo != 0) {
        return 0;
    }

    return resultado;
}
````

## Fase 2 · El Científico (10 min)

Redacta el contrato **C-R-E-O** (Contexto, Rol, Especificación, Output) para pedirle
la misma función a la IA, y pega el resultado en la celda de `ia_sesion10_c.c`. La
firma debe ser idéntica a la del contrato; si la IA la cambia, corrígela antes de
compilar.


````python
%%writefile ia_sesion10_c.c

/*
 * ia_sesion10_c.c
 * Fase 2: El Cientifico
 * Mini-evaluador de expresiones
 */

#include "api_sesion10.h"
#include <ctype.h>
#include <stddef.h>

/* Estructura para controlar el analisis */
typedef struct {
    const char *texto;
    int posicion;
    int error;
} Analizador;

/* Declaracion de funciones */
static long leer_expresion(Analizador *a);

/* Ignorar espacios en blanco */
static void saltar_espacios(Analizador *a) {
    while (isspace((unsigned char)a->texto[a->posicion])) {
        a->posicion++;
    }
}

/* Leer numeros y parentesis */
static long leer_factor(Analizador *a) {
    long valor = 0;
    int signo = 1;

    saltar_espacios(a);

    /* Revisar signo */
    if (a->texto[a->posicion] == '-') {
        signo = -1;
        a->posicion++;
    } else if (a->texto[a->posicion] == '+') {
        a->posicion++;
    }

    saltar_espacios(a);

    /* Revisar parentesis */
    if (a->texto[a->posicion] == '(') {
        a->posicion++;

        valor = leer_expresion(a);

        saltar_espacios(a);

        if (a->texto[a->posicion] != ')') {
            a->error = 2;
            return 0;
        }

        a->posicion++;

        return signo * valor;
    }

    /* Comprobar que exista un numero */
    if (!isdigit((unsigned char)a->texto[a->posicion])) {
        a->error = 2;
        return 0;
    }

    /* Leer los digitos */
    while (isdigit((unsigned char)a->texto[a->posicion])) {
        valor = valor * 10 +
                (a->texto[a->posicion] - '0');

        a->posicion++;
    }

    return signo * valor;
}

/* Resolver multiplicaciones y divisiones */
static long leer_termino(Analizador *a) {
    long resultado = leer_factor(a);
    char operador;

    while (a->error == 0) {
        saltar_espacios(a);

        operador = a->texto[a->posicion];

        if (operador != '*' && operador != '/') {
            break;
        }

        a->posicion++;

        long siguiente = leer_factor(a);

        if (a->error != 0) {
            return 0;
        }

        if (operador == '*') {
            resultado = resultado * siguiente;
        } else {
            if (siguiente == 0) {
                a->error = 1;
                return 0;
            }

            resultado = resultado / siguiente;
        }
    }

    return resultado;
}

/* Resolver sumas y restas */
static long leer_expresion(Analizador *a) {
    long resultado = leer_termino(a);
    char operador;

    while (a->error == 0) {
        saltar_espacios(a);

        operador = a->texto[a->posicion];

        if (operador != '+' && operador != '-') {
            break;
        }

        a->posicion++;

        long siguiente = leer_termino(a);

        if (a->error != 0) {
            return 0;
        }

        if (operador == '+') {
            resultado = resultado + siguiente;
        } else {
            resultado = resultado - siguiente;
        }
    }

    return resultado;
}

/* Funcion solicitada por el contrato */
long evaluar_expresion(const char *fuente, int *error) {
    Analizador a;
    long resultado = 0;

    if (error != NULL) {
        *error = 0;
    }

    /* Verificar la cadena */
    if (fuente == NULL) {
        if (error != NULL) {
            *error = 2;
        }

        return 0;
    }

    /* Inicializar analizador */
    a.texto = fuente;
    a.posicion = 0;
    a.error = 0;

    /* Evaluar expresion */
    resultado = leer_expresion(&a);

    saltar_espacios(&a);

    /* Verificar caracteres sobrantes */
    if (a.texto[a.posicion] != '\0' && a.error == 0) {
        a.error = 2;
    }

    if (error != NULL) {
        *error = a.error;
    }

    if (a.error != 0) {
        return 0;
    }

    return resultado;
}
````

## Fase 3 · El Auditor (10 min)

El harness de abajo **no aprueba nada por sí solo**: ejecuta tu función caso por caso,
la compara contra el valor esperado e imprime `OK` o `FALLA`. Léelo antes de ejecutarlo:
los casos te dicen qué se está midiendo.


````python
%%writefile harness_sesion10_c.c
/* harness_sesion10_c.c - auditoria automatica de la Sesion 10
 *
 * Compilacion:
 *   gcc -std=c11 -Wall -Wextra -Werror harness_sesion10_c.c artesano_sesion10_c.c -o test_artesano
 *
 * El harness NO aprueba nada por si solo: cada linea se calcula
 * ejecutando tu implementacion y comparandola contra el valor esperado.
 * Si algun caso falla, el programa termina con codigo de salida 1.
 */
#include <stdio.h>
#include <string.h>
#include "api_sesion10.h"

static int probadas = 0;
static int fallidas = 0;

static void verificar_int(const char *caso, long esperado, long obtenido) {
    int ok = (esperado == obtenido);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12ld obtenido: %-12ld %s\n",
           probadas, caso, esperado, obtenido, ok ? "OK" : "FALLA");
}

int main(void) {
    printf("=== AUDITORIA SESION 10 - Mini-evaluador de expresiones (Hito 1) ===\n");
    printf("Caso                               esperado:              obtenido:              resultado\n");

    { int e=0; long r=evaluar_expresion("2+3*4",&e); (void)e; verificar_int("\"2+3*4\"", (long)(14), (long)(r)); }
    { int e=0; long r=evaluar_expresion("(2+3)*4",&e); (void)e; verificar_int("\"(2+3)*4\"", (long)(20), (long)(r)); }
    { int e=0; long r=evaluar_expresion("10/0",&e); (void)r; verificar_int("\"10/0\" -> error 1", (long)(1), (long)(e)); }
    { int e=0; long r=evaluar_expresion("2+",&e); (void)r; verificar_int("\"2+\" -> error 2", (long)(2), (long)(e)); }
    { int e=0; long r=evaluar_expresion("  7 ",&e); (void)e; verificar_int("\"  7 \"", (long)(7), (long)(r)); }
    { int e=0; long r=evaluar_expresion("5-10",&e); (void)e; verificar_int("\"5-10\"", (long)(-5), (long)(r)); }

    printf("\nRESULTADO: %d/%d pruebas superadas\n",
           probadas - fallidas, probadas);
    if (fallidas != 0) {
        printf("Aun hay %d caso(s) sin resolver. Revisa tu implementacion.\n",
               fallidas);
        return 1;
    }
    printf("TODO EN VERDE. Documenta tu auditoria en auditorias/auditoria_sesion10.md" "\n");
    return 0;
}
````
